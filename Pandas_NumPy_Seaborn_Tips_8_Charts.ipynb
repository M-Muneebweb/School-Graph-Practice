## 3. Real `tips` dataset load karna

Seaborn ke `load_dataset()` function se built-in example dataset load karenge. Agar ye function kisi wajah se fail ho, to Pandas official Seaborn dataset repository ki CSV file read karne ki koshish karega.

**Yaad rahe:** ye internet se liya gaya existing dataset hai, hum khud se data generate nahi kar rahe.

```python
# Pehle Seaborn ke built-in dataset loader ko try karein.
try:
    tips = sns.load_dataset("tips")

# Agar built-in loader fail ho, to dataset ki CSV file Pandas se read karein.
except Exception as error:
    print("Seaborn loader se dataset load nahi hua; CSV source try ho raha hai.")
    tips_url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/tips.csv"
    tips = pd.read_csv(tips_url)

# Original dataset ko tips naam ke DataFrame mein store kiya gaya hai.
# Pehli 5 rows dekh kar data ka structure samjhein.
tips.head()
```

## 4. Data ko samajhna — Pandas basics

Visualization se pehle dataset ko inspect karna achi data-analysis practice hai.

- `shape`: rows aur columns ki tadaad
- `info()`: columns ke data types aur non-null values
- `describe()`: numerical columns ki summary statistics
- `isna().sum()`: missing values ki count
- `value_counts()`: categorical values ki frequency

**Use case:** Ye checks real projects mein CSV ya database se data milne ke baad kiye jate hain, taake analysis se pehle data ki structure aur quality samajh aa jaye.

```python
# Rows aur columns ki tadaad check karein.
print("Dataset shape (rows, columns):", tips.shape)

# Column names, data types aur non-null counts dekhein.
tips.info()

# Numerical columns ki statistical summary dekhein.
display(tips.describe())

# Har column mein missing values count karein.
display(tips.isna().sum())

# Day column mein har din ki observations count karein.
display(tips["day"].value_counts())
```

## 5. NumPy aur Pandas se aik useful derived column banana

Dataset mein `total_bill` aur `tip` columns hain. In se hum tip percentage calculate kar sakte hain:

\[
\text{tip percentage} = \frac{\text{tip}}{\text{total bill}} \times 100
\]

Yahan NumPy ka `np.where()` use hoga taake zero bill ki hypothetical situation mein division-by-zero se bacha ja sake. Original dataset ko overwrite karne ke bajaye hum uski copy par derived column add karenge.

```python
# Original DataFrame ki copy banayein, taake source data unchanged rahe.
tips = tips.copy()

# Tip percentage calculate karein.
# np.where condition true ho to formula apply hota hai, warna 0 assign hota hai.
tips["tip_pct"] = np.where(
    tips["total_bill"] > 0,
    (tips["tip"] / tips["total_bill"]) * 100,
    0
)

# Naya column dekhne ke liye selected columns ki first rows display karein.
display(tips[["total_bill", "tip", "tip_pct", "day", "size"]].head())
```

# Chart 1 — Line Chart

## Line chart kya dikhata hai?

Line chart points ko line se connect karta hai. Ye ordered values ke sath kisi measure ka pattern ya trend dikhane ke liye useful hai.

### Common use cases
- Time ke sath daily/monthly sales trend
- Temperature ka weekly pattern
- Party size barhne par average bill ya tip ka change
- Website visits ka date-wise trend

**Is example mein:** Hum har party size ke liye average tip calculate kar ke line plot karenge.

**Important note:** Tips dataset mein exact visit dates nahi hain. Is liye yahan line chart time-series nahi hai; party size ke ordered groups ke darmiyan average tip ka pattern dikhata hai.

```python
# Har table size ke liye average tip aur record count calculate karein.
avg_tip_by_party = (
    tips.groupby("size", as_index=False)
        .agg(
            average_tip=("tip", "mean"),
            number_of_tables=("tip", "count")
        )
        .sort_values("size")
)

# Figure ka size set karein.
plt.figure(figsize=(9, 5))

# X-axis par party size aur Y-axis par average tip plot karein.
plt.plot(
    avg_tip_by_party["size"],
    avg_tip_by_party["average_tip"],
    marker="o",
    linewidth=2,
    label="Average tip"
)

# Title aur axis labels add karein.
plt.title("Average Tip by Party Size")
plt.xlabel("Number of People at the Table")
plt.ylabel("Average Tip")

# X-axis par har observed party size ko tick banayein.
plt.xticks(avg_tip_by_party["size"])

# Legend, grid aur layout set karein.
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

# Aggregated values bhi display karein, taake chart ke peeche data samajh aaye.
display(avg_tip_by_party)
```

# Chart 2 — Box Plot

## Box plot kya dikhata hai?

Box plot distribution ka summary show karta hai:
- Box ke andar line: **median**
- Box ka start: **25th percentile (Q1)**
- Box ka end: **75th percentile (Q3)**
- Box ki height: **IQR = Q3 − Q1**
- Whiskers: aam tor par 1.5 × IQR range ke andar values
- Whiskers ke bahar points: potential outliers

### Common use cases
- Different days ki bill distribution compare karna
- Departments ki salary distribution compare karna
- Exam scores ka spread samajhna
- Delivery time mein unusual values identify karna

**Is example mein:** Hum restaurant bills ko weekdays ke hisaab se compare karenge.

**Note:** Outlier lazmi data-entry error nahi hota. Wo genuine large bill bhi ho sakta hai.

```python
# Weekdays ka logical order define karein.
day_order = ["Thur", "Fri", "Sat", "Sun"]

# Figure ka size set karein.
plt.figure(figsize=(9, 5))

# Har din ke total bill ki distribution ka box plot banayein.
sns.boxplot(
    data=tips,
    x="day",
    y="total_bill",
    order=day_order,
    showfliers=True
)

# Title aur labels add karein.
plt.title("Restaurant Bill Distribution by Day")
plt.xlabel("Day")
plt.ylabel("Total Bill")

# Layout aur chart display karein.
plt.tight_layout()
plt.show()
```

# Chart 3 — Histogram

## Histogram kya dikhata hai?

Histogram numerical data ko intervals (bins) mein divide karta hai. Har bar batata hai ke us range mein kitni observations aati hain.

### Common use cases
- Restaurant bills ki ranges aur frequency
- Students ke marks ki distribution
- Customer age distribution
- Delivery durations ki frequency

**Is example mein:** Hum dekhte hain ke restaurant bills kis amount range mein zyada common hain.

**Histogram vs Bar chart:** Histogram numerical ranges ke liye hota hai; bar chart aam tor par separate categories compare karta hai.

```python
# Figure ka size set karein.
plt.figure(figsize=(9, 5))

# Total bill ki values ka histogram banayein.
sns.histplot(
    data=tips,
    x="total_bill",
    bins=20,          # Bill amounts ko 20 intervals mein divide karein.
    kde=True,         # Distribution ki smooth density curve bhi show karein.
    edgecolor="black"
)

# Title aur axis labels add karein.
plt.title("Distribution of Restaurant Bills")
plt.xlabel("Total Bill")
plt.ylabel("Number of Tables")

# Chart display karein.
plt.tight_layout()
plt.show()
```

# Chart 4 — Bar Chart

## Bar chart kya dikhata hai?

Bar chart categories ke darmiyan kisi value ka comparison karta hai. Bar ki height/length us category ki value represent karti hai.

### Common use cases
- Weekdays ke darmiyan average sales compare karna
- Products ki total revenue compare karna
- Different branches ke customers compare karna
- Survey categories ki counts compare karna

**Is example mein:** Har din ki average tip compare hogi.

**Tip:** Bar chart mein mean, median, sum ya count—jo metric question ke liye useful ho—wo choose ki ja sakti hai.

```python
# Har din ki average tip calculate karein.
avg_tip_by_day = (
    tips.groupby("day", observed=True, as_index=False)["tip"]
        .mean()
)

# Figure ka size set karein.
plt.figure(figsize=(9, 5))

# Bar chart mein days ka order manually set karein.
sns.barplot(
    data=avg_tip_by_day,
    x="day",
    y="tip",
    order=day_order
)

# Har bar ke upar average value annotate karein.
for position, value in enumerate(
    avg_tip_by_day.set_index("day").reindex(day_order)["tip"]
):
    if pd.notna(value):
        plt.text(position, value, f"{value:.2f}", ha="center", va="bottom")

# Title aur labels add karein.
plt.title("Average Tip by Day")
plt.xlabel("Day")
plt.ylabel("Average Tip")

# Chart show karein.
plt.tight_layout()
plt.show()
```

# Chart 5 — Pie Chart

## Pie chart kya dikhata hai?

Pie chart aik total ko slices mein divide karta hai. Har slice total mein category ka percentage/share show karti hai.

### Common use cases
- Total expenses ka category-wise share
- Survey responses ka proportion
- Total revenue mein product categories ka contribution
- Weekdays par collect hui tips ka share

**Is example mein:** Har din collect hui total tips ka overall tips mein percentage share show hoga.

**Kab avoid karein?** Bohat zyada categories hon, ya slices ki values bohat qareeb hon, to pie chart mushkil ho sakta hai. Aisi surat mein bar chart zyada readable hota hai.

```python
# Har din ki total tips calculate karein.
total_tip_by_day = (
    tips.groupby("day", observed=True)["tip"]
        .sum()
        .reindex(day_order)
        .dropna()
)

# Figure create karein.
plt.figure(figsize=(8, 8))

# Total tips ko day-wise slices mein show karein.
plt.pie(
    total_tip_by_day,
    labels=total_tip_by_day.index,
    autopct="%1.1f%%",       # Har slice ka percentage show karein.
    startangle=90,
    counterclock=False
)

# Chart ko title dein aur circle shape maintain karein.
plt.title("Share of Total Tips Collected on Each Day")
plt.axis("equal")
plt.show()
```

# Chart 6 — Bubble Chart

## Bubble chart kya dikhata hai?

Bubble chart scatter plot jaisa hota hai, lekin bubble ka size teesri numerical variable ko show karta hai.

### Common use cases
- Countries: GDP vs life expectancy, bubble size = population
- Stores: revenue vs profit, bubble size = customer count
- Products: price vs sales, bubble size = units sold
- Restaurant tables: bill vs tip, bubble size = party size

**Is example mein:**
- X-axis = total bill
- Y-axis = tip
- Bubble size = table par logon ki tadaad
- Color = Lunch/Dinner

**Interpretation:** Bubble ki position bill aur tip batati hai, jabke bubble ka size party size ko represent karta hai. Bubble size ko compare karte waqt scale ko bhi dhyan mein rakhein.

```python
# Figure ka size set karein.
plt.figure(figsize=(10, 6))

# Bubble chart create karne ke liye scatter plot use karein.
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    size="size",            # Bubble size party size se linked hai.
    hue="time",             # Bubble color Lunch/Dinner ko show karta hai.
    sizes=(40, 450),        # Bubble sizes ka visual range.
    alpha=0.65,
    edgecolor="black"
)

# Chart title aur axis labels.
plt.title("Bubble Chart: Bill, Tip and Party Size")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

# Legend aur layout set karein.
plt.legend(title="Meal / Party Size", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

# Chart 7 — Heatmap

## Heatmap kya dikhata hai?

Heatmap table ki numerical values ko colors se represent karta hai. Is se patterns aur high/low values aik nazar mein samajh aa sakti hain.

### Common use cases
- Numerical variables ka correlation
- Confusion matrix
- Day/time ke hisaab se activity
- Different categories ka performance matrix

**Is example mein:** `total_bill`, `tip`, `size` aur `tip_pct` ka correlation matrix show hoga.

### Correlation ko kaise samjhein?
- `+1` ke qareeb: strong positive linear relationship
- `-1` ke qareeb: strong negative linear relationship
- `0` ke qareeb: weak linear relationship

**Important:** Correlation causation prove nahi karti. Correlation sirf variables ke linear association ka measure hai.

```python
# Heatmap ke liye selected numerical columns choose karein.
numeric_columns = ["total_bill", "tip", "size", "tip_pct"]

# Selected columns ka Pearson correlation calculate karein.
correlation_matrix = tips[numeric_columns].corr()

# Figure ka size set karein.
plt.figure(figsize=(8, 6))

# Correlation values ko heatmap ke colors mein show karein.
sns.heatmap(
    correlation_matrix,
    annot=True,          # Har cell mein correlation number likhein.
    fmt=".2f",           # Values ko 2 decimal places tak format karein.
    cmap="coolwarm",     # Positive/negative values ko contrasting colors dein.
    center=0,            # Zero ko color scale ka center rakhein.
    vmin=-1,
    vmax=1,
    linewidths=0.5
)

# Title aur chart display.
plt.title("Correlation Heatmap of Tips Dataset")
plt.tight_layout()
plt.show()
```

# Chart 8 — Scatter Plot

## Scatter plot kya dikhata hai?

Scatter plot har row ko x-y coordinate par aik point ke taur par show karta hai. Ye do numerical variables ka relationship, clusters aur unusual observations dekhne mein madad karta hai.

### Common use cases
- Study hours vs exam marks
- Marketing spend vs revenue
- House size vs house price
- Restaurant bill vs tip

**Is example mein:** Total bill aur tip ke darmiyan relationship dekhenge. Point ka color smoking status show karega.

**Interpretation:** Agar points left se right upar ki taraf pattern banate hon, to positive association ka signal ho sakta hai. Is ka matlab ye nahi ke aik variable doosre ka cause hai.

```python
# Figure ka size set karein.
plt.figure(figsize=(10, 6))

# Har record ko point ki tarah plot karein.
sns.scatterplot(
    data=tips,
    x="total_bill",
    y="tip",
    hue="smoker",       # Color smoker status ko represent karega.
    style="time",       # Marker style Lunch/Dinner ko represent karega.
    s=75,
    alpha=0.75
)

# Title aur labels add karein.
plt.title("Relationship Between Total Bill and Tip")
plt.xlabel("Total Bill")
plt.ylabel("Tip")

# Legend show karein.
plt.legend(title="Smoker / Meal Time")
plt.tight_layout()
plt.show()
```

## 6. Quick comparison — Kaunsa chart kab use karein?

| Chart | Best for | Tips dataset example |
|---|---|---|
| Line chart | Ordered groups ka trend/pattern | Party size vs average tip |
| Box plot | Distribution, median, potential outliers | Day-wise total bill |
| Histogram | Numerical frequency distribution | Total bill ranges |
| Bar chart | Categories ka comparison | Average tip by day |
| Pie chart | Total ka percentage share | Total tips by day |
| Bubble chart | 3 numerical variables | Bill, tip, party size |
| Heatmap | Matrix aur correlation | Numeric-column correlation |
| Scatter plot | 2 numerical variables ka relationship | Bill vs tip |
